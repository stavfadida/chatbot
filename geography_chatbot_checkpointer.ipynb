# Geography Chatbot with LangGraph

A small geography-only chatbot for students. Enter `stop` to end the conversation.

```python
import os
from typing import Annotated, TypedDict

from google.colab import userdata
from langchain.chat_models import init_chat_model
from langchain_core.messages import AnyMessage, HumanMessage, SystemMessage
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.graph import END, START, StateGraph
from langgraph.graph.message import add_messages

os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
model = init_chat_model("openai:gpt-6-luna")

SYSTEM = SystemMessage(
    content=(
        "You are a helpful chatbot for students. Answer questions about geography only. "
        "If asked about another topic, politely explain that you can help only with geography."
    )
)
```

```python
class ChatState(TypedDict):
    messages: Annotated[list[AnyMessage], add_messages]

def chatbot(state: ChatState):
    reply = model.invoke([SYSTEM] + state["messages"])
    return {"messages": [reply]}

builder = StateGraph(ChatState)
builder.add_node("chatbot", chatbot)
builder.add_edge(START, "chatbot")
builder.add_edge("chatbot", END)
graph = builder.compile(checkpointer=InMemorySaver())
config = {"configurable": {"thread_id": "1"}}
```

```python
while True:
    text = input("You: ")
    if text.strip().lower() == "stop":
        print("Chat ended.")
        break

    result = graph.invoke(
        {"messages": [HumanMessage(content=text)]},
        config,
    )
    print("Bot:", result["messages"][-1].content)
```
